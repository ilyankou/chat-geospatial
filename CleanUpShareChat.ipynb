# Clean up SHareChat

- Dataset - https://huggingface.co/datasets/tucnguyen/ShareChat/tree/main
- Paper - https://arxiv.org/abs/2512.17843

```python
import pandas as pd
```

```python
# takes a few min

dfs = []
models = ['clude', 'grok', 'perplexity', 'gemini', 'chatgpt']

for model in models:
    dfs.append(
        pd.read_csv(
            f'sharechat/chatgpt_results_final_language_filtered.csv'
        ).query(
            'detected_language_final == "English"'
        ).query(
            'role == "user"'
        ).query(
            'message_index == 1' # Let's look at the first message from the user only
        )
    )
    
data = pd.concat(dfs)
```

```python
data
```

```python
print(data.shape)

data = data[ data.plain_text.str.len() <= 140 ]
data = data.drop_duplicates(subset=['plain_text'])

print(data.shape)
```

```python
data.to_parquet('interim/sharechat.en.len140.parquet')
```
